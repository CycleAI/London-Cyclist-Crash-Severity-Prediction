```python
import pandas as pd

# 1. Load the data
print("Loading data...")
df = pd.read_csv('merged_collision_data.csv', low_memory=False)

# 2. Define the "Dual Schema" merge plan
# We take the primary column and fill missing values from the shortened 'secondary' column
merge_plan = [
    ('police_force', 'police_force', 'police_for'),
    ('day_of_week', 'day_of_week', 'day_of_wee'),
    ('light_conditions', 'light_conditions', 'light_cond'),
    ('weather_conditions', 'weather_conditions', 'weather_co'),
    ('road_surface_conditions', 'road_surface_conditions', 'road_surfa'),
    ('collision_severity', 'collision_severity', 'collisio_3'),
    ('speed_limit', 'speed_limit', 'speed_limi'),
    ('local_authority_district', 'local_authority_district', 'local_auth'),
    ('urban_or_rural_area', 'urban_or_rural_area', 'urban_or_r'),
    ('number_of_vehicles', 'number_of_vehicles', 'number_of_'),
    ('number_of_casualties', 'number_of_casualties', 'number_o_1')
]

print("Merging split columns...")
for target_col, pri, sec in merge_plan:
    if pri in df.columns and sec in df.columns:
        # Merge values
        df[target_col] = df[pri].fillna(df[sec])
        # Drop the now-redundant secondary column
        df = df.drop(columns=[sec])

# 3. Create high-value predictive features
print("Creating time-based features...")
# Convert date to standard format
df['date'] = pd.to_datetime(df['date'], dayfirst=True, errors='coerce')

# Extract Hour (identified as a high-impact factor)
df['hour'] = pd.to_datetime(df['time'], errors='coerce').dt.hour

# 4. Save the clean dataset
output_filename = 'cleaned_collision_data.csv'
df.to_csv(output_filename, index=False)

print(f"✅ Success! Cleaned file saved as: {output_filename}")
print(f"Final dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")
```
