```python
import requests
import geopandas as gpd

# --- URLs and file names ---
gpkg_url = "https://data.london.gov.uk/download/e55pw/9502cdec-5df0-46e3-8aa1-2b5c5233a31f/London_Boroughs.gpkg"
gpkg_file = "London_Boroughs.gpkg"
geojson_file = "london_boroughs.geojson"

# --- Step 1: Download the GeoPackage ---
print("Downloading London borough boundaries...")
response = requests.get(gpkg_url, timeout=60)

if response.status_code != 200:
    raise Exception(f"Download failed with status code {response.status_code}")

with open(gpkg_file, "wb") as f:
    f.write(response.content)

print(f"Downloaded: {gpkg_file}")

# --- Step 2: Load with GeoPandas ---
print("Reading GeoPackage...")
gdf = gpd.read_file(gpkg_file)

print("Columns available:", gdf.columns.tolist())
print(f"Number of boroughs: {len(gdf)}")

# --- Step 3: Convert CRS to WGS84 (required for Leaflet) ---
print("Converting CRS to EPSG:4326...")
gdf = gdf.to_crs(epsg=4326)

# --- Step 4: Save as GeoJSON ---
print("Saving as GeoJSON...")
gdf.to_file(geojson_file, driver="GeoJSON")

print(f"Done! File saved as: {geojson_file}")
```
